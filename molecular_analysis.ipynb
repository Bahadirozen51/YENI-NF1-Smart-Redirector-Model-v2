# NF1-Smart-Redirector-Model: Moleküler Analiz ve Görselleştirme

Bu notebook, `analyze_structure.py` simülatör modülünü kullanarak **Seed: 185615162** parametrelerine göre LNP boyut dağılımlarını (DLS) ve SRX-RNA01 doz-yanıt (Hill) eğrilerini simüle edip grafikleştirir.

```python
import numpy as np
import matplotlib.pyplot as plt
from analyze_structure import NF1SmartRedirectorModel

# Simülatör modelimizi baslatalim
model = NF1SmartRedirectorModel(seed=185615162, lock_distance_angstrom=2.85)
```

## 1. DLS Boyut Dağılım Grafiği (grafik1.png)
LNP Yapay Kapsülleme kararlılığını optimize etmek için hidrodinamik çap spektrumu simülasyonu.

```python
sizes, intensity = model.simulate_dls_size_distribution(target_radius_nm=90.0, pdi=0.15)

plt.figure(figsize=(8, 5))
plt.plot(sizes, intensity, color='#2bc473', lw=2.5, label='LNP Yoğunluk Spektrumu')
plt.axvline(x=90.0, color='red', linestyle='--', label='Hedef Çap (90 nm)')
plt.title('DLS Hidrodinamik Boyut Dağılımı (Faz 2 Optim.)')
plt.xlabel('Parçacık Çapı (nm)')
plt.ylabel('Bağıl Yoğunluk (%)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.savefig('grafik1.png', dpi=300)
plt.show()
```

## 2. Doz-Yanıt Hill Eğrisi Grafiği (grafik2.png)
Hipotetik IC50 = 0.45 nM parametresine göre SRX-RNA01 transkriptinin mutant hücre hatlarındaki inhibisyon profili.

```python
doses, inhibition = model.simulate_hill_dose_response()

plt.figure(figsize=(8, 5))
plt.semilogx(doses, inhibition, color='#1f77b4', lw=2.5, label='SRX-RNA01 İnhibisyon Eğrisi')
plt.axvline(x=0.45, color='orange', linestyle='--', label='Hipotetik IC50 (0.45 nM)')
plt.title('Doz-Yanıt Profil Simülasyonu (Hill Denklemi)')
plt.xlabel('Doz Konsantrasyonu (nM - Logaritmik)')
plt.ylabel('Onkoprotein İnhibisyonu (%)')
plt.grid(True, which='both', alpha=0.3)
plt.legend()
plt.savefig('grafik2.png', dpi=300)
plt.show()
```
